```python
import pandas as pd
import numpy as np
import openpyxl
import math
```

```python
def iferror(value, default):
    try:
        return eval(value)
    except:
        return default
    
kurs = 16449
value_kuartal = 4
```

```python
Pendapatan_dari_pembiayaan_konsumen_before = pd.read_excel("hasil/Data_Laporan_2024_Audit.xlsx" ,sheet_name="lr_now", index_col="kode entitas")
Pendapatan_dari_pembiayaan_konsumen_before
```

```python
persen_saham = pd.read_excel("2021 Q2 SAHAM - Copas .xlsx").drop([0,1,2]).reset_index(drop=True)
persen_saham.columns = persen_saham.iloc[0]
persen_saham = persen_saham.drop(0)
persen_saham = persen_saham.rename(columns={"Kode entitas":"kode entitas"}).set_index("kode entitas")

persen_saham
```

```python
lr_now = pd.read_excel("./hasil/Data_Laporan_2025_TW1.xlsx", sheet_name="lr_now", index_col="kode entitas")
pk_now = pd.read_excel("./hasil/Data_Laporan_2025_TW1.xlsx", sheet_name="pk_now", index_col="kode entitas")
ak_now = pd.read_excel("./hasil/Data_Laporan_2025_TW1.xlsx", sheet_name="ak_now", index_col="kode entitas")
lr_prev = pd.read_excel("./hasil/Data_Laporan_2025_TW1.xlsx", sheet_name="lr_prev", index_col="kode entitas")
pk_prev = pd.read_excel("./hasil/Data_Laporan_2025_TW1.xlsx", sheet_name="pk_prev", index_col="kode entitas")
ak_prev = pd.read_excel("./hasil/Data_Laporan_2025_TW1.xlsx", sheet_name="ak_prev", index_col="kode entitas")
stock_info = pd.read_excel("./hasil/Data_Laporan_2025_TW1.xlsx", sheet_name="stock_info", index_col="kode entitas")
```

```python
key = pd.read_excel("hasil/Data_Laporan_2024_Audit.xlsx", sheet_name="gen_info")["kode entitas"]

gen_info = pd.read_excel("hasil/Data_Laporan_2024_Audit.xlsx", sheet_name="gen_info")
gen_info = gen_info.set_index("kode entitas")
gen_info
```

```python
ringkasan_saham = pd.read_excel("Ringkasan Saham.xlsx").drop("No", axis=1)
ringkasan_saham = ringkasan_saham.rename(columns= {"Kode Saham":"kode entitas"})
ringkasan_saham = ringkasan_saham.set_index("kode entitas")
ringkasan_saham
```

```python
daftar_saham = pd.read_excel("Daftar Saham.xlsx", sheet_name="Sheet1")
daftar_saham.columns = ["no","kode entitas","nama entitas","ipo","jumlah saham","papan pencatatan"]
daftar_saham = daftar_saham.drop("no", axis=1).set_index("kode entitas")
daftar_saham['jumlah saham']=(
    daftar_saham['jumlah saham']
    .str.replace('.', '', regex=False)
    .astype(int)
)
daftar_saham
```

```python
main_df = pd.DataFrame({
    "kode entitas": [],
    "nama entitas": [],
    "sektor": [],
    "subsektor": [],
    "currency": [],
    "value currency": [],
    "satuan": [],
    "value satuan" : [],
    "auditor": [],
    "ipo": [],
    "jumlah saham": [],
    "% publik": [],
    "saham publik": [],
    "market cap in mio": [],
    "now": [],
    "target": [],
    "safety": [],
    "per": [],
    "ey": [],
    "peru": [],
    "bv": [],
    "pbv": [],
    "pbvx": [],
    "cfps": [],
    "tobins'q": [],
    "ev ebitda": [],
    "asset": [],
    "liabilits": [],
    "ekuitas": [],
    "income yoy": [],
    "income qoq": [],
    "profit": [],
    "ocf yoy": [],
    "roe": [],
    "roa": [],
    "current ratio": [],
    "cash ratio": [],
    "der": [],
    "tie": [],
    "npm": [],
    "gpm": [],
    "dividen": [],
    "dy": [],
    "ocf": [],
    "icf": [],
    "fcf": [],
    "laba rugi": [],
    "dar": [],
    "likuiditas": [],
    "underwriting": [],
    "kecukupan dana": [],
    "premi yoy": [],
    "beban klaim": [],
    "retensi": []
})

main_df["kode entitas"] = key
main_df = main_df.set_index("kode entitas")
main_df
```

```python
main_df.update(daftar_saham)
```

```python
main_df["sektor"] = gen_info["sektor"]
main_df["subsektor"] = gen_info["subsektor"]
main_df["currency"] = gen_info["mata uang pelaporan"]
main_df["satuan"] = gen_info["pembulatan yang digunakan dalam penyajian jumlah dalam laporan keuangan"]
main_df["now"] = stock_info["penutupan"]
main_df["% publik"] = persen_saham ["% Publik"]
main_df["peru"] = persen_saham["PERU"]
main_df["value currency"] = np.where(main_df["currency"].str.lower() == 'rupiah / idr', 1, kurs)
main_df["value satuan"] = np.where(main_df["satuan"].str.lower() == 'jutaan / in million', 1000000, 
                          np.where(main_df["satuan"].str.lower() == 'ribuan / in thousand', 10000, 
                                   1))

main_df["saham publik"] = round(main_df["jumlah saham"]*main_df["% publik"]).apply(lambda x: math.floor(x) if pd.notna(x) else pd.NA)
main_df["market cap in mio"] = round(main_df["jumlah saham"]*main_df["now"]/1000000).apply(lambda x: math.floor(x) if pd.notna(x) else pd.NA)


main_df
```

```python
main_df["target"] = (main_df["peru"] * lr_now["laba (rugi) per saham dasar dari operasi yang dilanjutkan"] * value_kuartal * main_df["value currency"]).fillna(0)
main_df["safety"] = ((main_df["target"] - main_df["now"]) / main_df["now"])
main_df["per"] = main_df["now"]/(lr_now["laba (rugi) per saham dasar dari operasi yang dilanjutkan"] * value_kuartal * main_df["value currency"])
main_df["ey"] = lr_now["laba (rugi) per saham dasar dari operasi yang dilanjutkan"] * value_kuartal * main_df["value currency"]/main_df["now"]
main_df["bv"] = (pk_now["jumlah ekuitas yang diatribusikan kepada pemilik entitas induk"]*main_df["value satuan"]*main_df["value currency"])/main_df["jumlah saham"]
main_df["pbv"] = main_df["now"] / main_df["bv"]
main_df["pbvx"] = persen_saham["PBVX"]
main_df["cfps"] = (ak_now["jumlah arus kas bersih yang diperoleh dari (digunakan untuk) aktivitas operasi"] *
                   main_df["value satuan"] *
                   main_df["value currency"] /
                   main_df["jumlah saham"])
main_df["tobins\'q"] = ((main_df["market cap in mio"] * 1000000) +
                        (pk_now["jumlah liabilitas"] * pk_now["jumlah aset tidak lancar"]*pk_now["persediaan lancar"])/
                        (pk_now["jumlah aset"]* pk_now["jumlah aset tidak lancar"] * pk_now["persediaan lancar"]))
main_df['ev ebitda'] = main_df.apply(
    lambda row: iferror(lambda: (
        ((row["market cap in mio"] * 1_000_000) + pk_now["jumlah liabilitas"] - row["kas dan setara kas"]) /
        round(lr_now["jumlah laba (rugi) sebelum pajak penghasilan"] * row["value satuan"] * row["value currency"])
    ), 0), axis=1)

"""---------------------------------------------------"""
main_df["asset"] = ((pk_now["jumlah aset"]-pk_prev["jumlah aset"]) / abs(pk_prev["jumlah aset"]))
main_df["liabilits"] = ((pk_now["jumlah liabilitas"] - pk_prev["jumlah liabilitas"]) / abs(pk_prev["jumlah liabilitas"]))
main_df["ekuitas"] = ((pk_now["jumlah ekuitas yang diatribusikan kepada pemilik entitas induk"] - pk_prev["jumlah ekuitas yang diatribusikan kepada pemilik entitas induk"]) / abs(pk_prev["jumlah ekuitas yang diatribusikan kepada pemilik entitas induk"]))
main_df["income yoy"] = ((lr_now["penjualan dan pendapatan usaha"] - lr_prev["penjualan dan pendapatan usaha"]) / abs(lr_prev["penjualan dan pendapatan usaha"]))
main_df["profit"] = ((lr_now["laba (rugi) yang dapat diatribusikan ke entitas induk"] - lr_prev["laba (rugi) yang dapat diatribusikan ke entitas induk"]) / abs(lr_prev["laba (rugi) yang dapat diatribusikan ke entitas induk"]))
main_df["ocf yoy"] = ((ak_now["jumlah arus kas bersih yang diperoleh dari (digunakan untuk) aktivitas operasi"] - ak_prev["jumlah arus kas bersih yang diperoleh dari (digunakan untuk) aktivitas operasi"]) / abs(ak_prev["jumlah arus kas bersih yang diperoleh dari (digunakan untuk) aktivitas operasi"]))
main_df["roe"] = (lr_now["laba (rugi) yang dapat diatribusikan ke entitas induk"] / abs(pk_now["jumlah ekuitas yang diatribusikan kepada pemilik entitas induk"]))
main_df["roa"] = (lr_now["laba (rugi) yang dapat diatribusikan ke entitas induk"] / abs(pk_now["jumlah aset"]))
main_df["current ratio"] = (pk_now["jumlah aset lancar"] / abs(pk_now["jumlah liabilitas jangka pendek"]))
main_df["cash ratio"] = (pk_now["kas dan setara kas"] / abs(pk_now["jumlah liabilitas jangka pendek"]))
main_df["der"] = (pk_now["jumlah liabilitas"] / abs(pk_now["jumlah ekuitas yang diatribusikan kepada pemilik entitas induk"]))
#main_df["tie"] = (lr_now["jumlah laba (rugi) sebelum pajak penghasilan"] / abs(lr_now["beban keuangan"]))
main_df["npm"] = (lr_now["laba (rugi) yang dapat diatribusikan ke entitas induk"] / abs(lr_now["penjualan dan pendapatan usaha"]))
main_df["gpm"] = (lr_now["jumlah laba bruto"] / abs(lr_now["penjualan dan pendapatan usaha"]))
main_df["dividen"] = (ak_now["pembayaran dividen dari aktivitas pendanaan"] *
                      main_df["value currency"] * main_df["value satuan"] / 
                      main_df["jumlah saham"])
main_df["dy"] = (main_df["dividen"] / main_df["now"])
main_df["income qoq"] = ((lr_now["penjualan dan pendapatan usaha"] - Pendapatan_dari_pembiayaan_konsumen_before["penjualan dan pendapatan usaha"]) / Pendapatan_dari_pembiayaan_konsumen_before["penjualan dan pendapatan usaha"])

# <------------------------------------------------------------------------------->
main_df["ocf"] = ak_now["jumlah arus kas bersih yang diperoleh dari (digunakan untuk) aktivitas operasi"].apply(
    lambda x: '1' if x > 0 else ('-1' if x < 0 else '0')
)

main_df["icf"] = ak_now["jumlah arus kas bersih yang diperoleh dari (digunakan untuk) aktivitas investasi"].apply(
    lambda x: '1' if x > 0 else ('-1' if x < 0 else '0')
)

main_df["fcf"] = ak_now["jumlah arus kas bersih yang diperoleh dari (digunakan untuk) aktivitas pendanaan"].apply(
    lambda x: '1' if x > 0 else ('-1' if x < 0 else '0')
)
# <------------------------------------------------------------------------------->

main_df["laba rugi"] = (lr_now["jumlah laba (rugi) sebelum pajak penghasilan"] / (pk_now["jumlah ekuitas yang diatribusikan kepada pemilik entitas induk"] + pk_prev["jumlah ekuitas yang diatribusikan kepada pemilik entitas induk"])/  2)
main_df['dar'] = (pk_now["jumlah liabilitas"] / pk_now["jumlah aset"])
main_df['likuiditas'] = (pk_now['jumlah liabilitas'] / pk_now["jumlah ekuitas yang diatribusikan kepada pemilik entitas induk"])
main_df["underwriting"] = ((lr_now["pendapatan dari premi asuransi"] - lr_now["beban klaim"]) / lr_now["pendapatan dari premi asuransi"])
main_df["kecukupan dana"] = (pk_now["jumlah ekuitas yang diatribusikan kepada pemilik entitas induk"] / pk_now["jumlah aset"])
main_df["premi yoy"] = ((lr_now["pendapatan dari premi asuransi"] - lr_prev["pendapatan dari premi asuransi"]) / lr_prev["pendapatan dari premi asuransi"])
main_df["beban klaim"] = (lr_now["beban klaim"] / lr_now["pendapatan dari premi asuransi"])
main_df["retensi"] = ((lr_now["pendapatan dari premi asuransi"] + lr_now["premi reasuransi"]) / lr_now["pendapatan dari premi asuransi"])



main_df
```
