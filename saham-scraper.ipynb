```python
from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from bs4 import BeautifulSoup
import time
import pandas as pd
import tqdm
```

```python
def extract_shareholder_table_with_selenium(url):
    
    chrome_options = Options()
    chrome_options.add_argument("--headless")
    chrome_options.add_argument("--disable-gpu")
    chrome_options.add_argument("user-agent=Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36")
    
    driver = webdriver.Chrome(options=chrome_options)
    
    try:
        driver.get(url)
        time.sleep(8)
        
        soup = BeautifulSoup(driver.page_source, 'html.parser')
        
        table = soup.find('table', {'id': 'table-shareholder'})
        
        if not table:
            raise Exception("Tabel tidak ditemukan")
        
        rows = table.find_all('tr')
        data = []
        
        for row in rows:
            cols = row.find_all(['th', 'td'])
            cols = [col.text.strip() for col in cols]
            data.append(cols)
        
        return data
    
    except Exception as e:
        print(f"Error: {e}")
        return None
    
```

```python
def shareholder_downloader ():
    hasil = {"nama entitas" : [],
             "jumlah saham" : [],
             "saham publik" : []}


    informasi = pd.read_excel("hasil/Data_Laporan_2025_TW1.xlsx", sheet_name="gen_info")
    informasi = informasi[["kode entitas", "nama entitas"]]
    informasi = informasi.apply(lambda x: x.str.lower() if x.dtype == 'object' else x)
    informasi['nama entitas'] = (
        informasi['nama entitas']
        .str.replace(r'pt|tbk|[^a-zA-Z0-9\s]', '', regex=True)  # Hapus PT/TBK dan simbol
        .str.strip()  # Hapus spasi berlebih
    )
    informasi["nama entitas"] = informasi["nama entitas"].str.replace(" ", "-")
    
    for index, row in informasi.iterrows():
        nama_entitas = row["nama entitas"]
        kode_entitas = row["kode entitas"]

        try :
            link = f"https://www.idnfinancials.com/id/{kode_entitas}/pt-{nama_entitas}-tbk"
            table_data = extract_shareholder_table_with_selenium(link)

            df = pd.DataFrame(table_data)
            df[1] = df[1].str.replace(r'\D+', '', regex=True).replace('', pd.NA).astype('Int64')
            jumlah_saham = df[1].sum()

            df = df[df[0].str.contains('Public', case=False, na=False)]
            saham_publik = df[1].sum()
            
            hasil["nama entitas"].append(nama_entitas)
            hasil["jumlah saham"].append(jumlah_saham)
            hasil["saham publik"].append(saham_publik)

            print(kode_entitas, " berhasil")

        except Exception as e:
            print(e)

    hasil = pd.DataFrame(hasil)
    return hasil
```
